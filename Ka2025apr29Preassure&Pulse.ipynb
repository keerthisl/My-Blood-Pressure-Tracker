```python
# health monitor, KA try desktop; 2025apr28 , ChatGpt ,  Plot Preassure & pulse seperately in same graph

# --- Load Libraries ---
import pandas as pd
import matplotlib.pyplot as plt

# --- Load and Clean Data ---
df = pd.read_csv('ka_daily_health.csv')

# Clean column names
df.columns = df.columns.str.strip()

# Convert 'Diastolic' and 'Pulse' to numeric
df['Diastolic'] = pd.to_numeric(df['Diastolic'], errors='coerce')
df['Pulse'] = pd.to_numeric(df['Pulse'], errors='coerce')

# Convert 'Time' column into DateTime format
df['DateTime'] = pd.to_datetime(df['Time'], format='%H:%M %d/%m/%Y', errors='coerce')

# --- Show Cleaned Data ---
print(df[['DateTime', 'Systolic', 'Diastolic', 'Pulse']])
print(df.dtypes)

# --- Plotting Systolic, Diastolic, and Pulse Together ---

plt.figure(figsize=(10, 6))

# Plot Systolic
plt.plot(df['DateTime'], df['Systolic'], marker='o', label='Systolic')

# Plot Diastolic
plt.plot(df['DateTime'], df['Diastolic'], marker='s', label='Diastolic')

# Plot Pulse
plt.plot(df['DateTime'], df['Pulse'], marker='x', label='Pulse')

# Titles and Labels
plt.title('Blood Pressure and Pulse Over Time')
plt.xlabel('Date and Time')
plt.ylabel('Measurements')

# Rotate X-axis labels
plt.xticks(rotation=45)

# Grid and Legend
plt.grid(True)
plt.legend()

# Fit layout
plt.tight_layout()

# Save the Plot
plt.savefig("Keerthi_Preassure_Pulse.png")

# Show the Plot
plt.show()
```

```python
# health monitor, KA try desktop; 2025apr28 , ChatGpt ,  Plot Preassure & pulse seperately

# --- Load Libraries ---
import pandas as pd
import matplotlib.pyplot as plt

# --- Load and Clean Data ---
df = pd.read_csv('ka_daily_health.csv')

# Clean column names
df.columns = df.columns.str.strip()

# Convert 'Diastolic' and 'Pulse' to numeric
df['Diastolic'] = pd.to_numeric(df['Diastolic'], errors='coerce')
df['Pulse'] = pd.to_numeric(df['Pulse'], errors='coerce')

# Convert 'Time' column into DateTime format
df['DateTime'] = pd.to_datetime(df['Time'], format='%H:%M %d/%m/%Y', errors='coerce')

# --- Show Cleaned Data ---
print(df[['DateTime', 'Systolic', 'Diastolic', 'Pulse']])
print(df.dtypes)

# --- Plotting Systolic and Pulse Separately ---
fig, axs = plt.subplots(2, 1, figsize=(8, 10))  # 2 rows, 1 column

# Plot Systolic
axs[0].plot(df['DateTime'], df['Systolic'], marker='o', color='blue')
axs[0].set_title('Systolic Blood Pressure Over Time')
axs[0].set_xlabel('Date and Time')
axs[0].set_ylabel('Systolic Pressure')
axs[0].tick_params(axis='x', rotation=45)
axs[0].grid(True)

# Plot Pulse
axs[1].plot(df['DateTime'], df['Pulse'], marker='x', color='orange')
axs[1].set_title('Pulse Over Time')
axs[1].set_xlabel('Date and Time')
axs[1].set_ylabel('Pulse Rate')
axs[1].tick_params(axis='x', rotation=45)
axs[1].grid(True)

plt.tight_layout()
plt.show()
```
